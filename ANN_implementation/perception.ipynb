------------- Perceptron -----------------

A Perceptron is the simplest form of an artificial neuron. It takes inputs, applies weights to them, adds a bias, and produces an output.


Perceptron = neural network ka basic building block.

Perceptrons → Neural Network → Deep Neural Network → Deep Learning

In Simple Way-

Perceptron = 1 neuron

Neural Network = many neurons

Deep Neural Network = many layers of neurons

Deep Learning = deep neural networks se complex problems solve karna

==================================================================================

ANN = Artificial Neural Network------------------

Input Layer → Hidden Layer(s) → Output Layer

--------------- propagation ----------------------

In Deep Learning, propagation means the flow of information through a neural network.

1. Forward Propagation

Information moves from input → hidden layers → output.

2. Backward Propagation (Backpropagation)

After getting the prediction, the network compares it with the actual answer and sends the error backward through the network to update the weights.

Input → Hidden → Output
                    ↓
                 Error
                    ↓
             Update Weights


Simple flow

Forward Propagation → Prediction → Calculate Loss → Backpropagation → Update Weights
