RNN= Recurrent Neural Network

RNN is a type of Neural Network designed for sequential data, where the order of data matters.

Simple example

Suppose you have a sentence:

"I am learning AI"

RNN reads it step-by-step:

I → am → learning → AI

At each step, RNN keeps some information from the previous step and uses it to understand the current input.

How it works


Input 1 → [RNN] → Hidden State
                    ↓
Input 2 → [RNN] → Hidden State
                    ↓
Input 3 → [RNN] → Hidden State
                    ↓
Input 4 → [RNN] → Output


The important idea is:

Current input + Previous information → Current output


```python

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense
import warnings
```

```python
sentences = [
 "I love this product",
 "This movie made me smile",
 "Service was friendly and quick",
 "Today felt bright and happy",
 "This is the best day",
 "Absolutely fantastic experience",
 "I enjoyed every single moment",
 "Great job, well done",
 "The food tasted delicious",
 "Totally recommend to everyone",
 "Very satisfied with results",
 "This worked better than expected",
 "Amazing quality and value",
 "Such a pleasant surprise",
 "I feel positive about this",
 "I hate this product",
 "This movie bored me",
 "Service was rude and slow",
 "Today was cold and lonely",
 "This is the worst day",
 "Terrible experience overall",
 "I regret buying this",
 "Very disappointed with results",
 "The food tasted awful",
 "Do not recommend this",
 "It broke after one use",
 "Not worth the money",
 "Utterly frustrating and annoying",
 "I feel negative about this",
 "Such a waste of time",
]
labels = [1]*15 + [0]*15
labels = np.array(labels)
```

```python
vocab_size = 2000
tok = Tokenizer(num_words = vocab_size,oov_token="<OOV>")
tok.fit_on_texts(sentences)
seq = tok.texts_to_sequences(sentences)
maxlen= max(len(s) for s in seq)
x= pad_sequences(seq,maxlen=maxlen,padding='post')
y= labels
```

```python
x[0]
```

```python
embed_dim= 16
rnn_units = 8
```

```python
inp = Input(
    shape=(maxlen,),
    dtype="int32",
    name="input"
)

X = Embedding(
    input_dim=vocab_size,
    output_dim=embed_dim,
   
    name="embed"
)(inp)

rnn = SimpleRNN(
    units=rnn_units,
    return_sequences=True,
    return_state=True,
    name="simple_rnn"
)

rnn_output, rnn_state = rnn(X)

out = Dense(
    1,
    activation="sigmoid",
    name="out"
)(rnn_state)

model = Model(
    inputs=inp,
    outputs=out
)

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
model.summary()
```

```python
history = model.fit(x,y,epochs=25,batch_size=8,verbose=1)
```
