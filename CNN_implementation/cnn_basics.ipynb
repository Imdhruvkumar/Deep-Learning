========================  CNN( Convolutional neural network )======================

CNN Steps----------------

Input Image
     ↓
1. Convolution
     ↓
2. ReLU Activation
     ↓
3. Pooling
     ↓
4. Flatten
     ↓
5. Fully Connected Layer
     ↓
6. Output

=====================================================================================================

1. Input Image

The image is given to the CNN.

Example:

Cat image → CNN

2. Convolution

A small filter/kernel moves over the image and detects features such as edges and shapes.

Image + Filter
      ↓
Feature Map


3. ReLU

ReLU adds non-linearity and removes negative values.

-2 → 0
 3 → 3
-5 → 0


4. Pooling

Pooling reduces the size of the feature map while keeping important information.

A common type is Max Pooling.

[1  5]
[3  2]

   ↓ Max Pooling

5


5. Flatten

The 2D feature maps are converted into a 1D vector.

Feature Maps
     ↓
Flatten
     ↓
[0.2, 0.8, 0.4, 0.1, ...]


6. Fully Connected Layer

The extracted features are given to normal neural-network layers to make the final decision.

7. Output

The model gives the prediction.

Cat image
   ↓
CNN
   ↓
Cat: 95%
Dog: 5%

CODE IMPLEMENTATION =========================================

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from sklearn.preprocessing import LabelEncoder , StandardScaler
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import Perceptron    # Used for simple linear classification tasks.

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from tensorflow.keras.models import Sequential     # Sequential lets you build a neural network layer-by-layer in Keras.

from tensorflow.keras.layers import Dense     #Dense makes the final predictions
from tensorflow.keras.layers import Conv2D     # Conv2D extracts features
from tensorflow.keras.layers import Flatten    # Flatten reshapes them

from tensorflow.keras.layers import MaxPooling2D     # MaxPooling2D reduces size
from tensorflow.keras.layers import Dropout          # Dropout prevents overfitting

from tensorflow.keras.utils import to_categorical     # converts numeric class labels into one-hot encoded format for training classification models
     
```

```python
df = pd.read_csv("mnist_train.csv", header=None)
df_test = pd.read_csv("mnist_test.csv", header=None)
```

```python
df.columns = ['label'] + [str(i) for i in range(784)]
df_test.columns = ['label'] + [str(i) for i in range(784)]
```

```python
# Separate input and target

X_train = df.drop('label', axis=1).values
y_train = df['label'].values

X_test = df_test.drop('label', axis=1).values
y_test = df_test['label'].values
```

```python
X_train = X_train.astype("float32")/ 255.0
X_test = X_test.astype("float32")/ 255.0
```

```python
X_train_img = X_train.reshape(-1,28,28)
X_test_img = X_test.reshape(-1,28,28)
```

```python
X_train_img
```

```python
y_train_cat = to_categorical(y_train,10)
y_test_cat = to_categorical(y_test,10)
```

```python
Perceptron = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10,activation='softmax')
])
```

```python
Perceptron.compile(optimizer="sgd",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
history_prep= Perceptron.fit(
    X_train_img,y_train_cat,
    epochs=5,
    batch_size = 32,
    validation_data = (X_test_img,y_test_cat),
    verbose =1
)
```

```python
prec_acc = Perceptron.evaluate(X_test_img,y_test_cat,verbose=0)[1]
```

```python
prec_acc
```

USING ANN============================

```python
ann = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(128,activation="relu"),
    Dense(64,activation="relu"),
    Dense(10,activation="softmax")
])
```

```python
ann.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
history_ann= ann.fit(
    X_train_img,y_train_cat,
    epochs=5,
    batch_size = 32,
    validation_data = (X_test_img,y_test_cat),
    verbose =1
)
```

```python
ann_acc = ann.evaluate(X_test_img,y_test_cat,verbose=0)[1]
```

```python
ann_acc
```

USING CNN=============================

```python
X_train_cnn = X_train.reshape(-1,28,28,1)
X_test_cnn = X_test.reshape(-1,28,28,1)
```

```python
cnn = Sequential([
    Conv2D(32,kernel_size=(3,3),activation="relu",input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),
    Conv2D(64,kernel_size=(3,3),activation="relu",input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dense(128,activation='relu'),
    Dropout(0.5),
    Dense(10,activation='softmax')

])
```

```python
cnn.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
history_cnn= cnn.fit(
    X_train_cnn,y_train_cat,
    epochs=5,
    batch_size = 32,
    validation_data = (X_test_cnn,y_test_cat),
    verbose =1
)
```

```python
cnn_acc = cnn.evaluate(X_test_cnn,y_test_cat,verbose=0)[1]
```
