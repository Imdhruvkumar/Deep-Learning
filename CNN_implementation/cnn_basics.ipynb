========================  CNN( Convolutional neural network )======================

CNN Steps----------------

Input Image
     ↓
1. Convolution
     ↓
2. ReLU Activation
     ↓
3. Pooling
     ↓
4. Flatten
     ↓
5. Fully Connected Layer
     ↓
6. Output

=====================================================================================================

1. Input Image

The image is given to the CNN.

Example:

Cat image → CNN

2. Convolution

A small filter/kernel moves over the image and detects features such as edges and shapes.

Image + Filter
      ↓
Feature Map


3. ReLU

ReLU adds non-linearity and removes negative values.

-2 → 0
 3 → 3
-5 → 0


4. Pooling

Pooling reduces the size of the feature map while keeping important information.

A common type is Max Pooling.

[1  5]
[3  2]

   ↓ Max Pooling

5


5. Flatten

The 2D feature maps are converted into a 1D vector.

Feature Maps
     ↓
Flatten
     ↓
[0.2, 0.8, 0.4, 0.1, ...]


6. Fully Connected Layer

The extracted features are given to normal neural-network layers to make the final decision.

7. Output

The model gives the prediction.

Cat image
   ↓
CNN
   ↓
Cat: 95%
Dog: 5%
